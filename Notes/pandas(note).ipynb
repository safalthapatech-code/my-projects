# **Concatenate**

```python
india=pd.DataFrame({
                    'city':['mumbai','delhi','bangalore'],
                    'temperature':[32,45,30],
                    'humidity':[80,60,78]
})
us=pd.DataFrame({
                 'city':['new york','chicago','orlando'],
                 'temperature':[21,14,35],
                 'humidity':[68,65,75]
})
df=pd.concat([india,us]) #combine india and us dataframe (we can combine as many as we wish)
print(df)
print('*'*50)
df=pd.concat([india,us],ignore_index=True) #ignore_index=True→ignore index for us dataframe and continues india's index
print(df)
```

```python
df=pd.concat([india,us],keys=['india','us'])  #put extra india,us index
print(df)
```

```python
df.loc['india']  #retrive india index data
```

```python
temperature_df=pd.DataFrame({
                            'city':['mumbai','delhi','bangalore'],
                            'temperature':[32,45,30]
})
windspeed_df=pd.DataFrame({
                            'city':['mumbai','delhi','banglore'],
                            'windspeed':[7,12,9]
})
df=pd.concat([temperature_df,windspeed_df],axis=1)
print(df)
```

```python
s=pd.Series(['sunny','dry','cloudy'],name='event') #create a column named 'event' with 'sunny','dry','cloudy' data
df=pd.concat([temperature_df,s],axis=1)
print(df)
```

# **Merge**

```python
temperature_df=pd.DataFrame({
                            'city':['delhi','bangalore','mumbai','up'],
                            'temperature':[32,45,30,30]
})
windspeed_df=pd.DataFrame({
                            'city':['mumbai','delhi','bangalore'],
                            'windspeed':[7,12,9]
})
df=pd.merge(temperature_df,windspeed_df,on='city',how='inner')
#if cities aren't in same position then concat fails but merge doesn't
# how='inner'→(temperature_df n windspeed_df),n→intersection
print(df)
print('*'*50)
df=pd.merge(temperature_df,windspeed_df,on='city',how='outer') # how='outer'→(temperature_df U windspeed_df)
print(df)
#how='left'→temperature_df
#how='Right'→windspeed_df
```

```python
df1=pd.DataFrame({
                    'city':['mumbai','delhi','bangalore'],
                    'temperature':[32,45,30],
                    'humidity':[80,60,78]
})
df2=pd.DataFrame({
                 'city':['bangalore','mumbai','delhi'],
                 'temperature':[21,14,35],
                 'humidity':[68,65,75]
})
df3=pd.merge(df1,df2,on='city')
print(df3)
```

# **Crosstab**

```python
df=pd.read_csv('crosstab.csv')
df
```

```python
pd.crosstab(df.Nationality,df.Handedness) #create crosstab table
#df.Nationality→rows,df.Handedness→column
#the below table tells us from bangladesh 2 are left handed and 0 are right handed and same for other rows
```

```python
pd.crosstab(df.Nationality,df.Handedness,margins=True) #margins=True→gives totali.i sum of left+right
```

```python
pd.crosstab([df.Nationality,df.Sex],df.Handedness,margins=True)
```

```python
pd.crosstab(df.Nationality,[df.Sex,df.Handedness],margins=True)
```

```python
pd.crosstab(df.Sex,df.Handedness,normalize='index') #normalize='index'→gives percentage
```

```python
pd.crosstab(df.Sex,df.Handedness,values=df.Age,aggfunc=np.average)
# in below table , average age of left handed female is 44.5,right handed is 31.0 and so on
```
