```python
#Importing necessary library
import numpy as np
X=np.array([1,1.5,2,2.5,3])
y=np.array([150,200,230,280,310])
#calculating means
mean_X=np.mean(X)
mean_y=np.mean(y)

#standard deviation
sd_X=np.std(X,ddof=1)
sd_y=np.std(y,ddof=1)

correlation=np.corrcoef(X,y)[0,1]

#calculating covarience
slope=(correlation*sd_y)/sd_X

#calculating intercept
intercept=mean_y-(slope*mean_X)

print(f"Slope: {slope}")
print(f"Intercept: {intercept}")

y_pred=slope*X+intercept
print('Predicted values=',y_pred)
```

```python
import pandas as pd
X=np.array([1,1.5,2,2.5,3])
y=np.array([150,200,230,280,310])
df=pd.DataFrame({'X':X,'y':y})
print(df)
#
```

```python
from scipy import stats

#perform linear regression using scipy
result=stats.linregress(X,y)

slope_scipy=result.slope
intercept_scipy=result.intercept
r_value_scipy=result.rvalue
p_value_scipy=result.pvalue
std_err_scipy=result.stderr

print(f"Slope (b1): {slope_scipy:.4f}")
print(f"Intercept (b0): {intercept_scipy:.4f}")
print(f"R-squared (R^2): {r_value_scipy**2:.4f}")
print(f"P-value: {p_value_scipy:.4f}")
print(f"Standard Error: {std_err_scipy:.4f}")

#predictions using scipy coefficients
y_pred_scipy=slope_scipy*X+intercept_scipy
print('Predicted values=',y_pred_scipy)
```
