```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import cross_val_score
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df=load_diabetes()
```

```python
# Load dataset
df = load_diabetes()

# Features and target
X = df.data
y = df.target
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# ------------------------------------------------
#  Split data: 80% Train, 20% Test
# ------------------------------------------------
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
models={"LinearRegression":LinearRegression(),
        "DecisionTreeRegressor":DecisionTreeRegressor(random_state=42),
        "RandomForestRegressor":RandomForestRegressor(random_state=42),
        "XGBRegressor":XGBRegressor(random_state=42)
        }
```

```python
cv_scores={}
for model_name,model in models.items():
   scores=cross_val_score(model,X_train,y_train,cv=5)
   cv_scores[model_name]=scores
   print(f'Training {model_name} with default parameters')
   print(f'{model_name} cross validation R^2:{np.mean(scores):.2f}')
   print('-'*50)

'''
R² score	Meaning
1.00	Perfect predictions
0.80	Very strong
0.60	Good
0.30	Some predictive power
0.00	No better than predicting the mean
< 0	Worse than the mean baseline
'''
```

```python
cv_scores
```

```python
#here best model is RandomForestRegressor ,second best is XGBRegressor and worst is DecisionTreeRegressor
best_model=LinearRegression()
best_model.fit(X_train,y_train)
```

```python
pred=best_model.predict(X_test)
```

```python
r2=r2_score(y_test,pred)
```

```python
print('R2 score=',r2)
'''
R² score	Meaning
1.00	Perfect predictions
0.80	Very strong
0.60	Good
0.30	Some predictive power
0.00	No better than predicting the mean
< 0	Worse than the mean baseline
'''
```
